```python
splits = {}

for seed in SEEDS:
    X_train, X_val, y_train, y_val = train_test_split(
        X_remaining,
        y_remaining,
        test_size=0.125,
        stratify=y_remaining,
        random_state=seed
    )

    splits[seed] = {
        "X_train": X_train,
        "X_val": X_val,
        "y_train": y_train,
        "y_val": y_val
    }

    print(f"Seed {seed}")
    print("Training:", X_train.shape)
    print("Validation:", X_val.shape)
    print("Test:", X_test.shape)
    print("--------------------")
```

```python
original_indices = np.arange(len(y))

remaining_idx, test_idx = train_test_split(
    original_indices,
    test_size=0.20,
    stratify=y,
    random_state=42
)

splits = {}

for seed in SEEDS:
    train_idx, val_idx = train_test_split(
        remaining_idx,
        test_size=0.125,
        stratify=y[remaining_idx],
        random_state=seed
    )

    splits[seed] = {
        "train_idx": train_idx,
        "val_idx": val_idx,
        "test_idx": test_idx,
        "X_train": X[train_idx],
        "X_val": X[val_idx],
        "y_train": y[train_idx],
        "y_val": y[val_idx]
    }

    print(f"\nSeed {seed}")

    print("Train:", len(train_idx))
    print("Validation:", len(val_idx))
    print("Test:", len(test_idx))

    assert len(set(train_idx) & set(val_idx)) == 0
    assert len(set(train_idx) & set(test_idx)) == 0
    assert len(set(val_idx) & set(test_idx)) == 0

    assert len(train_idx) + len(val_idx) + len(test_idx) == len(y)

    print("No overlap: PASSED")
    print("Total samples: PASSED")
```

```python
split_summary = []

for seed in SEEDS:
    split = splits[seed]

    for subset_name, indices in [
        ("Train", split["train_idx"]),
        ("Validation", split["val_idx"]),
        ("Test", split["test_idx"])
    ]:
        labels = y[indices]

        split_summary.append({
            "Seed": seed,
            "Subset": subset_name,
            "Samples": len(indices),
            "Percentage": round(len(indices) / len(y) * 100, 2),
            "Malignant (0)": int((labels == 0).sum()),
            "Benign (1)": int((labels == 1).sum())
        })

split_summary_df = pd.DataFrame(split_summary)

display(split_summary_df)
```

```python
from sklearn.preprocessing import StandardScaler

scaled_splits = {}
scalers = {}

for seed in SEEDS:
    split = splits[seed]

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(split["X_train"])
    X_val_scaled = scaler.transform(split["X_val"])

    scalers[seed] = scaler

    scaled_splits[seed] = {
        "X_train": X_train_scaled,
        "X_val": X_val_scaled
    }

    print(f"Seed {seed}")
    print("Scaled Train:", X_train_scaled.shape)
    print("Scaled Validation:", X_val_scaled.shape)
    print("--------------------")
```

```python
representations = {}

for seed in SEEDS:
    representations[seed] = {
        "Unscaled": {
            "X_train": splits[seed]["X_train"],
            "X_val": splits[seed]["X_val"]
        },
        "Scaled": {
            "X_train": scaled_splits[seed]["X_train"],
            "X_val": scaled_splits[seed]["X_val"]
        }
    }

    for rep_name, rep_data in representations[seed].items():
        print(
            f"Seed {seed} | {rep_name} | "
            f"Train: {rep_data['X_train'].shape} | "
            f"Validation: {rep_data['X_val'].shape}"
        )
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

model_configs = []

for depth in [2, 3, 5, 10, None]:
    model_configs.append({
        "algorithm": "Decision Tree",
        "parameter": depth
    })

for k in [1, 3, 5, 7, 11]:
    model_configs.append({
        "algorithm": "KNN",
        "parameter": k
    })

model_configs.append({
    "algorithm": "GaussianNB",
    "parameter": "Default"
})

configs_df = pd.DataFrame(model_configs)

display(configs_df)

print("Configurations:", len(model_configs))
print("Total fits:", len(model_configs) * 2 * len(SEEDS))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

def evaluate_model(model, X_train, y_train, X_val, y_val):

    train_pred = model.predict(X_train)
    val_pred = model.predict(X_val)

    # Breast Cancer: malignant = 0 (positive class)
    malignant_column = list(model.classes_).index(0)

    malignant_prob = model.predict_proba(X_val)[:, malignant_column]

    return {
        "train_accuracy": accuracy_score(y_train, train_pred),

        "train_macro_f1": f1_score(
            y_train, train_pred,
            average="macro",
            zero_division=0
        ),

        "val_accuracy": accuracy_score(y_val, val_pred),

        "val_macro_precision": precision_score(
            y_val, val_pred,
            average="macro",
            zero_division=0
        ),

        "val_macro_recall": recall_score(
            y_val, val_pred,
            average="macro",
            zero_division=0
        ),

        "val_macro_f1": f1_score(
            y_val, val_pred,
            average="macro",
            zero_division=0
        ),

        "val_roc_auc": roc_auc_score(
            (y_val == 0).astype(int),
            malignant_prob
        )
    }

print("Evaluation function is ready!")
```

```python
classifier_results = []
trained_models = {}

for seed in SEEDS:
    for rep_name, rep_data in representations[seed].items():

        X_train = rep_data["X_train"]
        X_val = rep_data["X_val"]

        y_train = splits[seed]["y_train"]
        y_val = splits[seed]["y_val"]

        for config in model_configs:

            algorithm = config["algorithm"]
            parameter = config["parameter"]

            if algorithm == "Decision Tree":
                model = DecisionTreeClassifier(
                    max_depth=parameter,
                    random_state=seed
                )
                config_id = f"DT_depth_{parameter}_{rep_name}"

            elif algorithm == "KNN":
                model = KNeighborsClassifier(
                    n_neighbors=parameter,
                    weights="uniform",
                    metric="euclidean"
                )
                config_id = f"KNN_k_{parameter}_{rep_name}"

            else:
                model = GaussianNB()
                config_id = f"GNB_default_{rep_name}"

            model.fit(X_train, y_train)

            metrics = evaluate_model(
                model, X_train, y_train, X_val, y_val
            )

            run_id = f"{config_id}_seed_{seed}"

            classifier_results.append({
                "dataset": "Breast Cancer",
                "run_id": run_id,
                "config_id": config_id,
                "algorithm": algorithm,
                "representation": rep_name,
                "parameter": str(parameter),
                "seed": seed,
                **metrics
            })

            trained_models[run_id] = model

classifier_runs_df = pd.DataFrame(classifier_results)

print("Total trained models:", len(classifier_runs_df))

display(classifier_runs_df.head(10))

print("\nModels per seed:")
print(classifier_runs_df.groupby("seed").size())

print("\nModels per representation:")
print(classifier_runs_df.groupby("representation").size())

assert len(classifier_runs_df) == 66

print("\nAll 66 classification runs completed!")
```

```python
metric_columns = [
    "train_accuracy",
    "train_macro_f1",
    "val_accuracy",
    "val_macro_precision",
    "val_macro_recall",
    "val_macro_f1",
    "val_roc_auc"
]

classifier_summary_df = (
    classifier_runs_df
    .groupby(
        ["dataset", "config_id", "algorithm",
         "representation", "parameter"],
        as_index=False
    )[metric_columns]
    .agg(["mean", "std"])
)

classifier_summary_df.columns = [
    "_".join(col).strip("_")
    for col in classifier_summary_df.columns
]

classifier_summary_df = classifier_summary_df.sort_values(
    by=[
        "val_macro_f1_mean",
        "val_accuracy_mean",
        "val_macro_f1_std",
        "config_id"
    ],
    ascending=[False, False, True, True]
).reset_index(drop=True)

print("Total configurations:", len(classifier_summary_df))

display(classifier_summary_df[
    [
        "config_id",
        "val_accuracy_mean",
        "val_macro_f1_mean",
        "val_macro_f1_std",
        "val_roc_auc_mean"
    ]
])
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score, silhouette_score

kmeans_models = {}
kmeans_results = []

for seed in SEEDS:
    for rep_name, rep_data in representations[seed].items():

        X_train_km = rep_data["X_train"]
        X_val_km = rep_data["X_val"]
        y_val_km = splits[seed]["y_val"]

        model = KMeans(
            n_clusters=2,
            random_state=seed,
            n_init=10
        )

        model.fit(X_train_km)

        val_clusters = model.predict(X_val_km)

        ari = adjusted_rand_score(y_val_km, val_clusters)

        if 2 <= len(set(val_clusters)) < len(val_clusters):
            silhouette = silhouette_score(X_val_km, val_clusters)
        else:
            silhouette = float("nan")

        run_id = f"KMeans_{rep_name}_seed_{seed}"

        kmeans_models[run_id] = model

        kmeans_results.append({
            "dataset": "Breast Cancer",
            "run_id": run_id,
            "representation": rep_name,
            "seed": seed,
            "n_clusters": 2,
            "train_inertia": model.inertia_,
            "val_ari": ari,
            "val_silhouette": silhouette
        })

kmeans_runs_df = pd.DataFrame(kmeans_results)

display(kmeans_runs_df)

print("Total K-Means fits:", len(kmeans_runs_df))
```

```python
kmeans_summary_df = (
    kmeans_runs_df
    .groupby(["dataset", "representation"], as_index=False)
    .agg(
        inertia_mean=("train_inertia", "mean"),
        inertia_std=("train_inertia", "std"),
        val_ari_mean=("val_ari", "mean"),
        val_ari_std=("val_ari", "std"),
        val_silhouette_mean=("val_silhouette", "mean"),
        val_silhouette_std=("val_silhouette", "std")
    )
)

display(kmeans_summary_df.round(4))

print("Total summary rows:", len(kmeans_summary_df))
```

```python
kmeans_test_results = []
test_cluster_predictions = {}

for rep_name in ["Unscaled", "Scaled"]:

    run_id = f"KMeans_{rep_name}_seed_42"
    model = kmeans_models[run_id]

    if rep_name == "Scaled":
        X_test_km = scalers[42].transform(X_test)
    else:
        X_test_km = X_test

    test_clusters = model.predict(X_test_km)

    test_cluster_predictions[rep_name] = test_clusters

    test_ari = adjusted_rand_score(
        y_test, test_clusters
    )

    if 2 <= len(np.unique(test_clusters)) < len(test_clusters):
        test_silhouette = silhouette_score(
            X_test_km, test_clusters
        )
    else:
        test_silhouette = np.nan

    cluster_counts = np.bincount(
        test_clusters,
        minlength=2
    )

    kmeans_test_results.append({
        "dataset": "Breast Cancer",
        "representation": rep_name,
        "seed": 42,
        "test_ari": test_ari,
        "test_silhouette": test_silhouette,
        "cluster_0_count": cluster_counts[0],
        "cluster_1_count": cluster_counts[1]
    })

kmeans_test_df = pd.DataFrame(kmeans_test_results)

display(kmeans_test_df.round(4))

print("Total K-Means test evaluations:", len(kmeans_test_df))
```

```python
contingency_tables = {}

for rep_name in ["Unscaled", "Scaled"]:

    clusters = test_cluster_predictions[rep_name]

    table = pd.crosstab(
        pd.Series(y_test, name="True Class"),
        pd.Series(clusters, name="Predicted Cluster")
    )

    table = table.reindex(
        index=[0, 1],
        columns=[0, 1],
        fill_value=0
    )

    table.index = ["Malignant (0)", "Benign (1)"]
    table.columns = ["Cluster 0", "Cluster 1"]

    contingency_tables[rep_name] = table

    print(f"\n{rep_name} K-Means — Contingency Table")
    display(table)

    print("Total samples:", table.to_numpy().sum())
```
